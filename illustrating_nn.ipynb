```python
import numpy as np
import scipy.io as sio
from sklearn.preprocessing import StandardScaler
from models import MouEtAl
import torch
import torch.nn as nn
from torchviz import make_dot
```

```python

# Parameters for Salinas dataset
input_size = 204       # Number of spectral bands
num_classes = 16       # Number of classes in Salinas dataset
patch_size = 7         # Patch size as you specified

# Select device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Instantiate model and move to device
model = MouEtAl(input_size=input_size, num_classes=num_classes, patch_size=patch_size)
model.to(device)

# Create dummy input tensor on the same device
# Shape: (batch_size, patch_size*patch_size, input_size)
dummy_input = torch.randn(1, patch_size * patch_size, input_size).to(device)

# Forward pass
output = model(dummy_input)
print(f"Output shape: {output.shape}")  # Should be (1, num_classes)

# Optional: visualize model graph with torchviz
try:
    dot = make_dot(output, params=dict(model.named_parameters()))
    dot.format = 'png'
    dot.render('MouEtAl_model_architecture')
    print("Model architecture graph saved as 'MouEtAl_model_architecture.png'")
except ImportError:
    print("torchviz not installed, skipping visualization.")
```
